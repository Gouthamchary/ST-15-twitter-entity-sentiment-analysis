```python
df = pd.read_csv("https://github.com/suhasmaddali/Twitter-Sentiment-Analysis/raw/refs/heads/main/train.csv")
```

```python
df=df[['text', 'sentiment']]
df
```

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
import re
def preprocess_text(text):
    if isinstance(text, float):
        return ""
    text = text.lower()  # Lowercase
    text = re.sub(r'http\S+|www.\S+', '', text)  # Remove URLs
    text = re.sub(r'[^a-zA-Z ]', '', text)  # Remove special characters & numbers 
    words = nltk.word_tokenize(text)  # Tokenization
    words = [word for word in words if word not in stopwords.words('english')]  # Remove stopwords
    return ' '.join(words)

tokens = []
for i in range(len(df)):
    tokens.append(preprocess_text(df['text'][i]))
```

```python
stop=stopwords.words('english')
stop_token=[]
for x in tokens:
  cl = [i for i in x if i not in stop]
  stop_token.append(cl)
```

```python
stop_token
```

```python
from nltk.stem import PorterStemmer
ps=PorterStemmer()
```

```python
stemedata =[]
for message in stop_token:
  st = [ps.stem(word) for word in message]
  stemedata.append(st)
```

```python
stemedata
```

```python
#apply pos_tagging on data
from nltk.tag import pos_tag
nltk.download('averaged_perceptron_tagger_eng')
```

```python
pos_token=[pos_tag(message) for message in stop_token]
pos_token
```

```python
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')
nltk.download('omw-1.4')
lemmatizer=WordNetLemmatizer()
```

```python
from nltk.corpus import wordnet

def get_wordnet_pos(treebank_tag):
    if treebank_tag.startswith('J'):
        return wordnet.ADJ
    elif treebank_tag.startswith('V'):
        return wordnet.VERB
    elif treebank_tag.startswith('N'):
        return wordnet.NOUN
    elif treebank_tag.startswith('R'):
        return wordnet.ADV
    else:
        return wordnet.NOUN
```

```python
get_wordnet_pos(pos_token[0][0][1])
```

```python
lem_data = []
for x in range(len(pos_token)):
  lem = [lemmatizer.lemmatize(pos_token[x][i][0],get_wordnet_pos(pos_token[x][i][1])) for i in range(len(pos_token[x]))]
  lem_data.append(lem)
```
